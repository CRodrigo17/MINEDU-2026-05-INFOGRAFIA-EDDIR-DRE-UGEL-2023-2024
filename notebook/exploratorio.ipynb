## Union de bases

```python
cols_bd1 = [
    'IDLista',
    'IDRegion',
    'DesRegion',
    'IDDreUgel',
    'DreUgel',
    'Documento',
    'DesCargo',
    'EstadoEvaluacion',
    'SeRealizoEvaluacion',
    'DesempenoEfectivo',
    'OR1_ptje',
    'OR2_ptje',
    'OR3_ptje',
    'LC1_ptje',
    'LC2_ptje',
    'LC3_ptje',
    'CS1_ptje',
    'CS2_ptje',
    'PRG_puntaje',
    'PCT_puntaje',
    'Puntaje_Final',
    'Condicion1',
    'Condicion2',
    'Condicion3',
    'CondicionFinal'
]

BD1_f = BD1[cols_bd1].copy()
```

```python
cols_bd2 = [
    'IDLista',
    'IDRegion',
    'Region',
    'IDDreUgel',
    'DreUgel',
    'DocumentoIdentidad',
    'Cargo',
    'EstadoEvaluacion',
    'SeRealizoEvaluacion',
    'DesempenoEfectivo',
    'OR1_ptje',
    'OR2_ptje',
    'OR3_ptje',
    'LC1_ptje',
    'LC2_ptje',
    'LC3_ptje',
    'CS1_ptje',
    'CS2_ptje',
    'PRG_puntaje',
    'PCT_puntaje',
    'Puntaje_Final',
    'Condicion1',
    'Condicion2',
    'Condicion3',
    'CondicionFinal'
]

BD2_f = BD2[cols_bd2].copy()
```

```python
BD2_f = BD2_f.rename(columns={
    "Region": "DesRegion",
    "DocumentoIdentidad": "Documento",
    "Cargo": "DesCargo"
})
```

```python
BD1_f["FUENTE"] = "DGP_JPG"
BD2_f["FUENTE"] = "DIR_UGEL"
```

```python
BD1_f.columns.equals(BD2_f.columns)
```

```python
BD_final = pd.concat(
    [BD1_f, BD2_f],
    ignore_index=True
)
```

```python
BD_final["CondicionFinal"].value_counts(dropna=False)
```
